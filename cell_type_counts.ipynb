```python
import pandas as pd
import matplotlib.pyplot as plt
import glob
import os
```

```python

# Path where your CSV files are located
path = './local_data/important_genes/'   # change this

# Collect all CSV files
files = glob.glob(os.path.join(path, "*description.csv"))
# print(len(files))

df_list = []
for f in files:
    temp = pd.read_csv(f)
    df_list.append(temp)

# Combine everything
df = pd.concat(df_list, ignore_index=True)

# Ensure cell_type is treated as category (optional but neat)
df["cell_type"] = df["cell_type"].astype("category")

# Create boxplot
plt.figure(figsize=(12, 6))
df.boxplot(column="count", by="cell_type", rot=45)

plt.title("Boxplot of Count per Cell Type Across Samples")
plt.suptitle("")   # removes default subtitle
plt.xlabel("Cell Type")
plt.ylabel("Count")

plt.tight_layout()
plt.show()
```
